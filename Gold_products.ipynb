```python
from pyspark.sql.functions import monotonically_increasing_id

df_products = spark.table(
    "databricks_catalog.silver.silver_products"
)

df_products_gold = df_products.withColumn(
    "DimProductKey",
    monotonically_increasing_id() + 1
)

df_products_gold.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("databricks_catalog.gold.dimproducts")
```
